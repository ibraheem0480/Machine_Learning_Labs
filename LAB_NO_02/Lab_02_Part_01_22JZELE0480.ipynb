#### How Random Seed Works (NumPy)

np.random.seed(x) resets the random number generator to a fixed starting point.

Using the same seed → produces the same random numbers every time.

Seed affects only the next random operations after it is called.

In Jupyter Notebook, each cell continues the same random sequence.

So, if you set the seed in one cell and generate numbers in another, the result will change.

To get identical output in different cells, set the seed again before each random operation.

```python
# Setting a random seed value of 24 to guarantee reproducible output distributions
np.random.seed(24)
a = np.random.rand(4)
a
```

```python
# Generating subsequent numbers without resetting the seed yields the next values in line
b = np.random.rand(4)
b
```

### 2.3 Broadcasting Operations
Broadcasting allows mathematical operators to compute values across arrays of differing dimensions without creating manual loop iterations.

```python
# Temporarily broadcasting a scalar addition of 100 to all elements inside the array
arr + 100
```

```python
# Setting up a 2x3 Matrix 'A' alongside a 1D row vector 'b' to show vector broadcasting
A = np.array([[1, 2, 3],
              [4, 5, 6]])

b = np.array([10, 20, 30])

# Array 'b' gets duplicated across matching dimensions down matrix 'A' automatically
print(A + b)
```

```python
# Verifying that addition order does not disrupt the broadcasting mechanics
print(b + A)
```

```python
# Computing element-wise division by scaling down all inputs inside 'arr' by 2
arr / 2
```

```python
# Demonstrating operational precedence: divides 100 by 2 first, then adds result to 'arr'
arr + 100 / 2
```

```python
# Applying an element-wise exponent calculation to square every value inside the array
arr ** 2
```

# 3 NumPy Statistical Functions

```python
import numpy as np

arr = np.array([4, 9, 2, 7, 5])
print("Array:", arr)

# Mean
print("Mean:", arr.mean())

# Maximum value
print("Max:", arr.max())

# Minimum value
print("Min:", arr.min())

# Index of maximum value
print("Argmax:", arr.argmax())

# Index of minimum value
print("Argmin:", arr.argmin())

# Standard deviation
print("Standard Deviation:", arr.std())

# Variance
print("Variance:", arr.var())

# Median
print("Median:", np.median(arr))

# Sum of all elements
print("Sum:", arr.sum())

# Cumulative sum
print("Cumulative Sum:", arr.cumsum())

# Sorting
print("Sorted Array:", np.sort(arr))
```
