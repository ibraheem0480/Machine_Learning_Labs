## Pandas Data Structures: Series vs DataFrame

**Series**: One-dimensional labeled array (like a single column).
**DataFrame**: Two-dimensional labeled data structure (like a table with multiple columns).

## GroupBy Operations

```python
# Example dataset for GroupBy
data = {'Company':['GOOG','GOOG','MSFT','MSFT','FB','FB'],
       'Person':['Sam','Charlie','Amy','Vanessa','Carl','Sarah'],
       'Sales':[200,120,340,124,243,350]}
df = pd.DataFrame(data)
```

```python
by_comp = df.groupby('Company')
by_comp.count()
```

```python
# Detailed statistical summary by group
by_comp.describe()
```
